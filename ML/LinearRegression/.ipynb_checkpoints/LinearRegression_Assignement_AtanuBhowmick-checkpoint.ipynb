```python
# Mapping the month column

df.month.replace({1:'jan',2:'feb',3:'mar',4:'apr',5:'may',6:'june',7:'july',8:'aug',9:'sep',10:'oct',11:'nov',12:'dec'},inplace = True)
```

```python
## Data Preparation
```

```python
# Return back to 0,1 value for year colum for liner regression

df.year.replace({"2018" : 0, "2019" : 1}, inplace = True)
```

```python
months_df = pd.get_dummies(df.month, drop_first = True, dtype = "int")
weekdays_df = pd.get_dummies(df.weekday, drop_first = True, dtype = "int")
weathersit_df = pd.get_dummies(df.weathersit, drop_first = True, dtype = "int")
seasons_df = pd.get_dummies(df.season,drop_first = True, dtype = "int")
```

```python
months_df.head()
```

```python
weekdays_df.head()
```

```python
weathersit_df.head()
```

```python
# Merging  the dataframe, with the dummy variable dataset. 

df_new = pd.concat([df, months_df, weekdays_df, weathersit_df, seasons_df], axis = 1)
```

```python
df_new.head()
```

```python
df_new.info()
```

```python
# Removing columns as already created dummy variable out of them

df_new.drop(['season', 'month', 'weekday', 'weathersit'], axis = 1, inplace = True)
```

```python
df_new.info()
```

```python
# Splitting the Data into Training and Testing Sets

np.random.seed(0)
df_train, df_test = train_test_split(df_new, train_size = 0.7, random_state = 100)
```

```python
# Get the shape of train set

df_train.shape
```

```python
# Get the shape of test set

df_test.shape
```

```python
# Using MinMaxScaler to Rescaling the features

scaler = MinMaxScaler()
```

```python
# See the head of dataset before scaling.

df_train.head()
```

```python
# Apply scaler() to all the columns except the 'yes-no' and 'dummy' variables

num_variables = ['temp','atemp','humidity','windspeed','cnt']
df_train[num_variables] = scaler.fit_transform(df_train[num_variables])
```

```python
# See the head after appying scaling.

df_train.head()
```

```python
# Describing the dataset

df_train.describe()
```

```python
# Visualize the heatmap

plt.figure(figsize = (25,25))
sns.heatmap(df_train.corr(), annot = True, cmap="RdYlGn")
plt.show()
```

```python
# Visualize cnt column via Scatter plot.

plt.figure(figsize = [6,6])
plt.scatter(df_train.temp, df_train.cnt)
plt.show()
```

```python
# Building the Linear Model

y_train = df_train.pop('cnt')
X_train = df_train
```

```python
# Recursive feature elimination 

lm = LinearRegression()
lm.fit(X_train, y_train)

rfe = RFE(lm, n_features_to_select = 15)
rfe = rfe.fit(X_train, y_train)
```

```python
# List of variables selected in top 15 list

list(zip(X_train.columns, rfe.support_, rfe.ranking_))
```

```python
# Selecting the selected variable via RFE in col list

cols = X_train.columns[rfe.support_]
print(cols)
```

```python
# Checking which columns has been rejected

X_train.columns[~rfe.support_]
```

```python
def get_vif(df):
    vif = pd.DataFrame()
    vif['Features'] = df.columns
    vif['VIF'] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]
    vif['VIF'] = round(vif['VIF'], 2)
    vif = vif.sort_values(by = "VIF", ascending = False)
    return vif 
```

```python
# Dataframe with RFE selected variables

X_train_rfe = X_train[cols]
```

```python
X_train_rfe.info()
```

```python
# Calculate VIF for RFE train set

get_vif(X_train_rfe)
```

```python
## Build Models
```

```python
# Building 1st linear regression model

X_train_model_1 = sm.add_constant(X_train_rfe)
lr_1 = sm.OLS(y_train,X_train_model_1).fit()
print(lr_1.summary())
```

```python
# As humidity shows high VIF values hence we can drop it
X_train_final = X_train_rfe.drop(['humidity'], axis = 1)

# Run the function to calculate VIF for the new model
get_vif(X_train_final)
```

```python
# Building 2nd linear regression model

X_train_model_2 = sm.add_constant(X_train_final)
model_2 = sm.OLS(y_train, X_train_model_2).fit()
print(model_2.summary())
```

```python
# We can drop nov variable as it has high p-value
X_train_final = X_train_final.drop(['nov'], axis = 1)

# Run the function to calculate VIF for the new model
get_vif(X_train_final)
```

```python
# Building 3rd linear regression model

X_train_model_3 = sm.add_constant(X_train_final)
model_3 = sm.OLS(y_train, X_train_model_3).fit()
print(model_3.summary())
```

```python
# We can drop jan variable as it has high p-value
X_train_final = X_train_final.drop(['jan'], axis = 1)

# Run the function to calculate VIF for the new model
get_vif(X_train_final)
```

```python
# Building 4th linear regression model

X_train_model_4 = sm.add_constant(X_train_final)
model_4 = sm.OLS(y_train, X_train_model_4).fit()
print(model_4.summary())
```

```python
# We can drop july variable as it has high p-value
X_train_final = X_train_final.drop(['july'], axis = 1)

# Run the function to calculate VIF for the new model
get_vif(X_train_final)
```

```python
# Building 5th linear regression model

X_train_model_5 = sm.add_constant(X_train_final)
model_5 = sm.OLS(y_train, X_train_model_5).fit()
print(model_5.summary())
```

```python
# Checking the parameters and their coefficient values
model_5.params
```

```python
X_train_model_5.head()
```

```python
y_train_pred = model_5.predict(X_train_model_5)
```

```python
# Plot the histogram of the error terms

fig = plt.figure()
sns.distplot((y_train - y_train_pred), bins = 20)
fig.suptitle('Error Terms', fontsize = 20) 
plt.xlabel('Errors', fontsize = 18)
```

```python
get_vif(X_train_final)
```

```python
plt.figure(figsize = (15,8))
sns.heatmap(X_train_final.corr(), annot = True, cmap="RdYlGn")
plt.show()
```

```python
# Linear relationship validation using CCPR plot
# Component and component plus residual plot

sm.graphics.plot_ccpr(model_5, 'temp')
plt.show()
```

```python
sm.graphics.plot_ccpr(model_5, 'sep')
plt.show()
```

```python
sm.graphics.plot_ccpr(model_5, 'windspeed')
plt.show()
```

```python
# Test set prediction
```

```python
 # Applying scaling on the test dataset

num_variables = ['temp', 'atemp', 'humidity', 'windspeed','cnt']
df_test[num_variables] = scaler.transform(df_test[num_variables])
df_test.head()
```

```python
df_test.describe()
```

```python
y_test = df_test.pop('cnt')
X_test = df_test
```

```python
col1 = X_train_final.columns

X_test = X_test[col1]
```

```python
# Adding constant variable to test dataframe
X_test_lm_6 = sm.add_constant(X_test)
```

```python
y_pred = model_5.predict(X_test_lm_6)
```

```python
r2 = r2_score(y_test, y_pred)
round(r2, 4)
```

```python
## Model Evaluation
```

```python
# Plotting y_test and y_pred to understand the spread

fig = plt.figure()
plt.scatter(y_test, y_pred)
plt.xlabel('y_test')
plt.ylabel('y_pred') 
```

```python
# Get the params

round(model_5.params, 4)
```

```python
# Calculating Adjusted-R^2 value for test data set.

adjusted_r2 = round(1 - (1 - r2) * (X_test.shape[0] - 1) / (X_test.shape[0] - X_test.shape[1] - 1), 4)
print(adjusted_r2)
```

```python
# Visualize the fit on the test data.

plt.figure()
sns.regplot(x = y_test, y = y_pred, fit_reg = True)
plt.title('Predction Graph', fontsize=20)
plt.xlabel('y_test')
plt.ylabel('y_pred')
plt.show()
```
