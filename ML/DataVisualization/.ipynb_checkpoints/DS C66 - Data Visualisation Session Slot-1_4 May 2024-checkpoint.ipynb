# Visualization with Matplotlib

>It is multi-platform data visualization library, built on NumPy arrays.

>It is a Python alternative for MATLAB users. The MATLAB-style tools are contained in the *pyplot* (plt) interface.

>In recent years, the interface and style of MPL have begun to show its age. Newer tools like ggplot and ggvis in R language, along with web visualization toolkits based on D3js and HTML5 canvas. These all often make MPL feel clunky and old-fashioned. Still, I'm of the opinion that we cannot ignore MPL's strength as it is a well-tested, cross-platform graphics engine.

```python
plt.figure(figsize=(4,3))
plt.plot(x,y, label="sine") # sine wave
plt.plot(x,z, label="cosine") # cosine wave
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine") # sine wave
plt.plot(x,z, label="cosine") # cosine wave
plt.legend()
plt.show()
```

```python
# Adjust the position for legend
plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine") # sine wave
plt.plot(x,z, label="cosine") # cosine wave
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
plt.legend(loc='upper right')
plt.show()
```

__Adjust the colors__

>color by name: blue, green, yellow, pink, etc

>color by code: rgbcmyk https://blog.thenounproject.com/wp-content/uploads/2022/06/01-RGB-vs-CMYK.png
        
>color as Grayscale between 0 and 1: 0 will be dark grey, 1 will be white
    
>color by Hex code: RRGGBB from 00 to FF eg: #FFDD55
        
>color as RGB tuple: values from 0 to 1 eg: (1.0,0.1,0.3)

```python
# Adjust the colors

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c="yellow") # sine wave
plt.plot(x,z, label="cosine", c="k") # cosine wave
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the colors

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c="steelblue") # sine wave
plt.plot(x,z, label="cosine", c="y") # cosine wave
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the colors

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c="0.7") # sine wave
plt.plot(x,z, label="cosine", c="#003366") # cosine wave
# https://www.w3schools.com/colors/colors_picker.asp
# https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.legend.html
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the colors

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c=(0,1.0,0.6)) # sine wave
plt.plot(x,z, label="cosine", c=(1.0,0,0)) # cosine wave
plt.legend(loc='upper right')
plt.show()
```

__Adjust the linestyle__

>solid "-" It is default anyways

>dashed "--"

>dashdot "-."

>dotted ":"

```python
# Adjust the linestyles

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c=(0,0,1.0), linestyle="solid") # sine wave
plt.plot(x,z, label="cosine", c=(1.0,0,0), linestyle="dashed") # cosine wave
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the linestyles

plt.figure(figsize=(5,4))
plt.plot(x,y, label="sine", c=(0,0,1.0), linestyle="dashdot") # sine wave
plt.plot(x,z, label="cosine", c=(1.0,0,0), linestyle="dotted") # cosine wave
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the color and linestyles at the same time using short code

plt.figure(figsize=(5,4))
plt.plot(x,y, "-g",label="sine" ) # solid green
plt.plot(x,z, "--c", label="cosine") # dashed cyan
plt.legend(loc='upper right')
plt.show()
```

```python
# Adjust the color and linestyles at the same time using short code

plt.figure(figsize=(5,4))
plt.plot(x,y, "-.k",label="sine" ) # dashdot black
plt.plot(x,z, ":r", label="cosine") # dotted red
plt.legend(loc='upper right')
plt.show()
```

## Understand the two types of Interfaces

>1. MATLAB style Interface

>2. Object Oriented style Interface

#### MATLAB style Interface



```python
x = np.linspace(0,10,100) # generate 100 uniformly spaced values between (0,10)
y = np.sin(np.linspace(0,10,100))
z = np.cos(np.linspace(0,10,100))
```

```python
plt.figure(figsize=(2,4))

plt.subplot(2,1,1)
plt.plot(x,y)

plt.subplot(2,1,2)
plt.plot(x,z)

plt.show()
```

```python
plt.figure(figsize=(4,2))

plt.subplot(1,2,1) # (rows, columns, panel number)
plt.plot(x,y)

plt.subplot(1,2,2)
plt.plot(x,z)

plt.show()
```

```python
plt.figure(figsize=(5,5))

plt.subplot(2,2,1) # (rows, columns, panel number)
plt.plot(x,y)

plt.subplot(2,2,2)
plt.plot(x,z)

plt.subplot(2,2,4)
plt.plot(x,np.tan(x))

plt.show()
```

##### Bar Chart - Plot sales across each product category

- A bar chart uses bars to show comparisons between categories of data.
- A bar graph will always have two axis. 
- One axis will generally have numerical values or measures, 
- The other will describe the types of categories being compared or dimensions.

```python
product_category = np.array(['Furniture', 'Technology', 'Office Supplies'])
sales = np.array ([4110451.90, 4744557.50, 3787492.52] )

plt.figure(figsize=(4,3))
plt.bar(product_category, sales)
plt.show()
```

```python
plt.figure(figsize=(4, 3))
# plotting bar chart and setting bar width to 0.5 and aligning it to center
plt.bar(product_category,
        sales,
        width=0.5,
        align='center',
        edgecolor='Orange',
        color='cyan')

# Adding and formatting title
plt.title("Sales Across Product Categories\n",
          fontdict={
              'fontsize': 20,
              'fontweight': "heavy",
              'color': 'Green'
          })
# weight or fontweight
# [ 'normal' | 'bold' | 'heavy' | 'light' | 'ultrabold' | 'ultralight']
plt.show()
```

##### Box and Whisker Chart - Sales across Countries and Product Categories

A Box and Whisker Plot (or Box Plot) is a convenient way of visually displaying the data distribution through their quartiles.
