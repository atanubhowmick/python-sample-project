```python
# Create a income segment
def incomeRange(income):
    if income <= 50000:
        return 'VL' # Very low income
    elif income >= 50001 and income <= 100000:
        return 'L' # low income
    elif income >= 100001 and income <= 200000: 
        return 'M' # Medium income
    elif income >= 200001 and income <= 300000: 
        return 'MTH' # Medium to High income
    elif income >= 300001 and income <= 500000: 
        return 'H' # High income
    elif income >= 500001 and income <= 800000: 
        return 'VH' # Very high income
    else:
        return 'UH' # Ultra high
```

```python
inp0['INCOME_RANGE'] = inp0['AMT_INCOME_TOTAL'].apply(lambda amt: incomeRange(amt))
```

```python
# Percentage of missing values for each column
allColsPercentage = inp1.isnull().sum() * 100.0 / inp1.shape[0]
allColsPercentage
```

```python
# Retrieving the columns for which missing percentage is more than 40%
missingCols = allColsPercentage[allColsPercentage.values > 40]
missingCols
```

```python
# Get the missing columns
missingColList = list(missingCols.index)
missingColList
```

```python
# Drop the missing columns
inp1.drop(labels = missingColList, axis = 1, inplace = True)
```

```python
colsNotRequired = ['NFLAG_LAST_APPL_IN_DAY', 
                   'HOUR_APPR_PROCESS_START', 
                   'WEEKDAY_APPR_PROCESS_START', 
                   'FLAG_LAST_APPL_PER_CONTRACT']
inp1.drop(labels = colsNotRequired, axis = 1, inplace = True)
```

```python
# Check for null values
inp1.isnull().sum()
```

```python
# Delete the rows where AMT_ANNUITY empty
inp1 = inp1[~inp1['AMT_ANNUITY'].isnull()]
```

```python
# Delete the rows where AMT_GOODS_PRICE empty
inp1 = inp1[~inp1['AMT_GOODS_PRICE'].isnull()]
```

```python
# Delete the rows where CNT_PAYMENT empty
inp1 = inp1[~inp1['CNT_PAYMENT'].isnull()]
```

```python
# Delete the rows where AMT_CREDIT empty
inp1 = inp1[~inp1['AMT_CREDIT'].isnull()]
```

```python
# Delete the rows where PRODUCT_COMBINATION empty
inp1 = inp1[~inp1['PRODUCT_COMBINATION'].isnull()]
```

```python
# Check for null values again
inp1.isnull().sum()
```

```python
inp1.info()
```

```python
# Remove outliers with 1.5 IQR rule
colsForOutlier = ['AMT_CREDIT', 
                    'AMT_ANNUITY',
                    'AMT_APPLICATION',
                    'AMT_GOODS_PRICE',
                    'CNT_PAYMENT',
                    'DAYS_DECISION']
for col in colsForOutlier: 
    q25 = inp1[col].quantile(0.25)
    q75 = inp1[col].quantile(0.75)
    iqr = q75 - q25
    low  = q25 - (1.5 * iqr)
    high = q75 + (1.5 * iqr)
    inp1 = inp1[~((inp1[col] < low) | (inp1[col] > high))]
```

```python
# Populate MONTH_DECISION columns
inp1['MONTH_DECISION'] = abs(inp1['DAYS_DECISION']//30)
```

```python
inp1.info()
```

```python
colToDelete = ['AMT_CREDIT', 'AMT_ANNUITY']
inp0.drop(labels = colToDelete, axis = 1, inplace = True)
```

```python
# Merging both the datasets into a single dataset
inp2 = pd.merge(inp1, inp0, on='SK_ID_CURR', how='left')
```

```python
inp2.shape
```

```python
inp2.isnull().sum()
```

```python
# Delete the rows where AMT_CREDIT empty
inp2 = inp2[~inp2['TARGET'].isnull()]
```

```python
ax = sns.countplot(x = 'NAME_PORTFOLIO', data = inp2)
ax.set_xticklabels(ax.get_xticklabels(), rotation = 45, horizontalalignment = 'right')
ax.set(xlabel='PORTFOLIO')
plt.show()
```
