```python
# Reverse by code
my_nums = [1, 2, 3, 4, 5, 6]
n = len(my_nums)
for i in range(n//2):
    temp = my_nums[i]
    my_nums[i] = my_nums[n - i - 1]
    my_nums[n - i - 1] = temp
my_nums
```

```python
# zip method
names = ["Alice", "Bob", "John"]
ages = [25, 20, 30, 40 , 50]

# zip() returns an iterator. We can wrap it in list() to view it all at once.
# For un-even length, zip drops the remaining elements
combined = list(zip(names, ages))
combined
```

```python
# # Merge two dictionaries - using union operator
dict1 = {'a': 1, 'b': 2}
dict2 = {'b': 3, 'c': 4}

# Creates a brand new dictionary
merged = dict1 | dict2
merged
```

```python
# Sort dict by values
scores = {'john': 85, 'jane': 92, 'dave': 78, 'alice': 95}

# Sort by value (index 1 of the tuple)
sorted_scores = dict(sorted(scores.items(), key=lambda item: item[1]))

print(sorted_scores)
# Output: {'dave': 78, 'john': 85, 'jane': 92, 'alice': 95}

# To sort in descending order, add reverse=True
desc_scores = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
print(desc_scores)
```

```python
# Count word frequency
text = "This is is Python Python Python"
freq = {}

for word in text.split():
    # dict.get(word, 0) returns the current count, or 0 if the word isn't in the dict yet
    freq[word] = freq.get(word, 0) + 1

print(freq)

# another way using counter
from collections import Counter
freq1 = dict(Counter(text.split()))
print(freq1)
```

```python
# Combine two dictionaries by adding values.
dict1 = {'a': 10, 'b': 20}
dict2 = {'a': 5, 'b': 8, 'c': 7}

dict3 = {k: dict1.get(k, 0) + dict2.get(k, 0) for k in dict1.keys() | dict2.keys()}
print(dict3)
```

## Set

```python
# Set Union
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# using Operator (union operator is applicable to dict also. But other operators are not applicable)
set_c = set_a | set_b
print(set_c)

# Using Method
set_d = set_a.union(set_b)
print(set_d)
```

```python
# Set Intersection
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# using Operator
set_c = set_a & set_b
print(set_c)

# Using Method
set_d = set_a.intersection(set_b)
print(set_d)
```

```python
# Set Difference
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# using Operator
set_c = set_a - set_b
print(set_c)

# Using Method
set_d = set_a.difference(set_b)
print(set_d)
```

```python
# Set Symmetric Difference
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

# using Operator
set_c = set_a ^ set_b
print(set_c)

# Using Method
set_d = set_a.symmetric_difference(set_b)
print(set_d)
```

```python
from collections import defaultdict

data = [
    ("HR", "Alice"),
    ("IT", "Bob"),
    ("HR", "John")
]

# Initialize with 'list' so any new key automatically starts with an empty []
grouped = defaultdict(list)

# Unpack the tuple directly in the loop for readability
for department, name in data:
    # No need to check if 'department' exists yet!
    grouped[department].append(name)

print(dict(grouped))
```

## Queue

```python
# Priority Queue - min heap
import heapq

# A heap in Python is just a regular list
min_heap = []

# 1. Push / Insert
heapq.heappush(min_heap, 10)
heapq.heappush(min_heap, 5)
heapq.heappush(min_heap, 20)
heapq.heappush(min_heap, 2)

print(min_heap)

smallest = heapq.heappop(min_heap)
print(smallest)
```

```python
# Priority Queue - max heap
import heapq

# Start with an empty standard list
max_heap = []

# Data to insert
values = [10, 20, 5, 40, 30]

# 1. PUSH: Multiply by -1 before pushing
for val in values:
    heapq.heappush(max_heap, -val)

# Internally, the heap looks like this: [-40, -30, -5, -10, -20]
# The smallest number is -40, so it sits at index 0.

# 2. POP: Multiply by -1 to get your max value back
largest = -heapq.heappop(max_heap)
print(f"Largest: {largest}")  # Output: Largest: 40

second_largest = -heapq.heappop(max_heap)
print(f"Second Largest: {second_largest}")
```

```python
# nth largest and smallest
import heapq

nums = [6, 10, 4, 2, 8, 1, 9, 7, 3, 5]

# 4th Largest
# Returns the top four - [10, 9, 8, 7], then we slice the last one
fourth_largest = heapq.nlargest(4, nums)[-1]
print(fourth_largest) # Output: 7

# 2nd Smallest
# Returns the bottom two - [1, 2], then we slice the last one
second_smallest = heapq.nsmallest(2, nums)[-1]
print(second_smallest) # Output: 2
```

```python
# deque
from collections import deque

# Initialize with an iterable
d = deque([2, 3])

# Append (Add to ends)
d.append(4)       # Right side: deque([2, 3, 4])
d.appendleft(1)   # Left side:  deque([1, 2, 3, 4])

# Pop (Remove from ends)
right = d.pop()       # Returns 4. deque is now: deque([1, 2, 3])
left = d.popleft()    # Returns 1. deque is now: deque([2, 3])
```
